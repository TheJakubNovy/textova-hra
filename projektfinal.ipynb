```python
inventar=[]
chybi=["doplnit palivo", "opravit padák", "opravit elektřinu"]
zivoty=100
import random
zije=True
vrstva=0
leva_chodba=0
prava_chodba=0
leva_mistnost=0
probouzeni=1
start_mistnost=0
suplik=1
skrin=1
modre_dvere=True
cervene_dvere=True
souboj_chodba=1
pravy_suplik=1
levy_suplik=1
prostredni_chodba=0
elektrina=0
suplik_mid_chodba=1
prava_mistnost=0
unikovy_modul=0
palivo=0
padak=0
kod=str(random.randint(1000, 9999))
#funkce zařizuje kontrolu vstupu a příkazy životy a inventář, 
#POZOR to že po jednom z příkazů se ve většině situací vypíše else(zadej 1, 2...) je zamýšlené aby to hráči připomnělo co má za možnosti
def oprava_int_input(prompt):
    vstup= input(prompt)
    vstup=vstup.strip()
    vstup=vstup.lower()
    try:
        return int(vstup) 
    except ValueError:
        if inv_health(vstup):
            return
        else:
            return 0
#funkce vypisuje inventář nebo životy pokud o to hráč požádá
def inv_health(input):
    if input=="inventář":
        print("tvůj inventář obsahuje:", ", ".join(inventar))
        return True
    elif input=="životy":
        print("máš",zivoty,"životů")
        return True
#vrací false pokud je hráč mrtvý
def check_hrac_zije():
    global zije
    if zivoty <= 0:
        zije=False
        return False
    return True

def probuzeni_def():
        global zije
        global zivoty
        global vrstva
        global probouzeni
        global start_mistnost
        while True:
            vstup=oprava_int_input("napiš 1 pro začátek hry")
            if vstup==1:
                print("probouzíš se na zemi, kolem tebe jsou rozházené věci, skoro nic si nepamatuješ, jen si pamatuješ na nějaký náraz")
                break
            elif vstup is not None:
                print("máš jen možnost napsat 1")
        while True:
            vstup=oprava_int_input("co uděláš? 1- ležíš dál na zemi a snažíš se vzpomenout si kde jsi, 2- zvedneš se a rozhlédneš se kolem sebe")
            if vstup==1:
                print ("vzpomínáš si na to že jsi astronaut a pracuješ na mezinárodní vesmírné stanici, jak dál ležíš tak ti najednou lupne v uších a začne se ti motat hlava")
                while True :
                    vstup=oprava_int_input(" 1. zůstáváš ležet 2. zvedneš se a podíváš se kolem,")
                    if vstup==1:
                        print("ležíš a snažíš se vzpomenout si na víc, dál ti praská v uších a potom se ti zatmívá před očima a omdlíváš,")
                        print("Umřel jsi na udušení, zkus to znovu")
                        zije=False
                        break
                    if vstup==2:
                        break
                    else:
                        print("napiš 1 nebo 2")
                if not zije :
                    break
            if vstup==2 and zije:
                print("vstáváš a hned vidíš že ve stěně je díra, má asi šířku palce a utíká skrz ní vzduch na druhou stranu kde jsou skrz díru vidět hvězdy")
                break
            else:
                print("napiš 1 nebo 2")
        while zije:
            if vrstva==1:
                break
            vstup1=oprava_int_input("co uděláš? 1. kašleš na to a jdeš se porozhlédnout okolo 2. zacpeš to prstem 3. utrhneš si kus oblečení a zacpeš to")
            if vstup1==1:
                print("před sebou vidíš dveře, jsou zamknuté na kód, jdeš k nim blíž ale blbě se ti dýchá, padáš k zemi a dusíš se")
                print("Umřel jsi na udušení, zkus to znovu")
                zije=False
                break
            elif vstup1==2:
                print("běžíš k díře a zacpeš jí prstem, prst tě bolí ale dočasně to jde vydržet")
                while zije:
                    if vrstva==1:
                        break
                    vstup=oprava_int_input("co uděláš? 1. rozhlédneš se kolem, 2. necháš prst na díře a uvidíš co se stane")
                    if vstup==1:
                        print("vidíš kus plechu který by se dal přiložit na stěnu, taky vidíš velký hadr")
                        while zije:
                            vstup=oprava_int_input("čím to zacpeš? 1. plechem 2. hadrem 3. nejdřív ke stěně přiložíš hadr a na to plech")
                            if vstup==1:
                                print("přikládáš plech, vzduch na chvíli přestane proudit, potom uslyšíš vrzání a plech se vcucne skrz díru, díra se zvětší a vytahuje tě do vesmíru za ní")
                                print("zemřel jsi na pobyt ve vakuu")
                                zije=False
                                break
                            if vstup==2:
                                print("dáváš hadr na díru ale tlak ho vcucne, nestihneš to už ničím zacpat než omdlíš")
                                print("zemřel jsi na udušení")
                                zije=False
                                break
                            if vstup==3:
                                print("drží to! žádný vzduch neuniká ")
                                vrstva=1
                                break
                            else:
                                print("zadej číslo 1 až 3")
                    if not zije:
                        break
                    if vrstva==1:
                        break
                    if vstup==2 and zije:
                        zivoty-=5
                        print("palec tě bolí, ztrácíš 5 životů, teď máš", zivoty, "životů")
                        zije=check_hrac_zije()
                        if not zije:
                            print("umřel jsi na vykrvácení")
                            break
                    else:
                        print("napiš 1 nebo 2")
                        continue
            
            elif vstup1==3:
                print("všímáš si že oblečení není z látky ale ze zvláštní gumy, na díře to nepropustně drží")
                break
            else:
                print("napiš 1, 2 nebo 3")
        if zije:
             print("sedáš si vyčerpaně na zem, a začínáš si jasně vzpomínat na to kde jsi, jsi astronaut na mezinárodní vesmírné stanici, vzpomínáš si že než jsi omdlel tak jí zasáhl roj asteroidů")
             print("přijde ti divné že cítíš gravitaci i přes to že jsi ve vesmíru, nevíš čím to může být,")
        probouzeni=0
        start_mistnost=1

#první místnost po probuzení a zacpání díry nebo vstupu do místnosti
def start_mistnost_def():
    print ("stojíš uprostřed místnosti a vidíš šuplík, skříň, modré dveře a červené dveře")
    global suplik
    global skrin
    global leva_chodba
    global start_mistnost
    global modre_dvere
    global zivoty
    global prava_chodba
    global cervene_dvere
    while True:
        vstup=oprava_int_input("kam jdeš 1. k šuplíku 2. k skříni 3. k modrým dveřím 4. k červeným dveřím")
        if vstup==1:
            if suplik==1:
                inventar.append("karta")
                inventar.append("lahev džusu (+ 50 životů)")
                print("našel jsi kartu a lahev džusu která ti při vypití doplní 50 životů ale dá se vypít jen při souboji, v inventáři teď máš", inventar)
                suplik=0
            elif suplik==0:
                print("v šuplíku nic není")
        elif vstup==2:
            if skrin==1:
                inventar.append("papírek s kódem:"+ kod)
                inventar.append("nůž")
                inventar.append("železná trubka")
                print("našel jsi papírek s kódem:", kod, "železnou trubku a nůž který ti přidává v souboji 5 útoku, v inventáři máš:", inventar)
                skrin=0
            elif skrin==0:
                print("ve skříni nic není, u sebe teď máš:", inventar)
        elif vstup==3:
            print("přicházíš ke dveřím, je vedle nich panel kam se dá zadat kód")
            while True:
                if modre_dvere:
                    vstup=oprava_int_input("můžeš 4 místný zadat kód nebo 1.odejít 2.zkusit dveře otevřít silou")
                    if str(vstup)==str(kod):
                        print("kód", kod, "byl správně, dveře se otevírají, za nimi je dlouhá chodba, vstupuješ do ní" )
                        leva_chodba=1
                        start_mistnost=0
                        return
                    else:
                        if vstup==1:
                            break
                        elif vstup==2:
                            if "železná trubka" in inventar:
                                print("podařilo se ti otevřít dveře je za nimi dlouhá chodba jdeš do ní")
                                start_mistnost=0
                                leva_chodba=1
                                modre_dvere=False
                                return
                            else:
                                utok=random.randint(1, 20)
                                if utok < 11:
                                    zivoty-=10
                                    print("dveře drží, od mlácení do dveří tě bolí ruka, ztrácíš 10 životů, teď máš", zivoty, "životů")
                                    check_hrac_zije()
                                    if not zije:
                                        print("zemřel jsi od mlácení do dveří, PROHRÁL JSI")
                                        return
                                elif 10<utok :
                                     print("podařilo se ti otevřít dveře je za nimi dlouhá chodba jdeš do ní")
                                     start_mistnost=0
                                     leva_chodba=1
                                     modre_dvere=False
                                     return
                        else:
                            print("špatný kód, zadej 1. nebo 2. nebo kód")
                else:
                    print("procházíš rozbitými dveřmi do dlouhé chodby")
                    start_mistnost=0
                    leva_chodba=1
                    return
        elif vstup==4:
            print("jdeš ke dveřím a zjišťuješ že se dají otevřít kartou")
            if cervene_dvere:
                if "karta" in inventar:
                    print("vkládáš kartu ze své kapsy a procházíš dveřmi do dlouhé chodby")
                    prava_chodba=1
                    start_mistnost=0
                    return
                else:
                    while True:
                        vstup=oprava_int_input("1.odcházíš 2.pokoušíš se dveře otevřít silou")
                        if vstup==1:
                            break
                        elif vstup==2:
                            if "železná trubka" in inventar:
                                print("podařilo se ti dveře otevřít, za nimi je dlouhá chodba, vstupuješ do ní")
                                start_mistnost=0
                                cervene_dvere=False
                                prava_chodba=1
                                return
                            else:
                                utok=random.randint(1, 20)
                                if utok<11:
                                    zivoty-=10
                                    print("dveře drží, od mlácení do dveří tě bolí ruka, ztrácíš 10 životů, teď máš", zivoty, "životů")
                                    check_hrac_zije()
                                    if not zije:
                                        print("zemřel jsi od mlácení do dveří, PROHRÁL JSI")
                                        return
                                elif 10<utok:
                                    print("podařilo se ti otevřít dveře je za nimi dlouhá chodba jdeš do ní")
                                    start_mistnost=0
                                    cervene_dvere=False
                                    prava_chodba=1
                                    return
                        else:
                            print("zadej 1 nebo 2")
            else:
                print("procházíš dveřmi do chodby")
                start_mistnost=0
                prava_chodba=1
                return
        else:
            print("napiš 1, 2, 3 nebo 4")

#souboj se slizem, proběhne v obou chodbách ale může proběhnout jen jednou za hru
#to že energie roste do nekonečna pokud hráč pořád odpočívá je zamýšlené 
def souboj_sliz():
    global zivoty
    global souboj_chodba
    global zije
    sliz_zije=True
    energie=15
    kolo=1
    sliz_zivoty=100
    print("v chodbě na zemi leží nějaká černá louže slizu, přibližuješ se k ní a najednou se začíná hýbat, formuje se do tvaru člověka a útočí na tebe")
    print("v této hře fungují souboje tak že máš 15 energie a za každý útok spotřebuješ 5, energie se dá doplnit tak že místo útoku odpočíváš, doplníš jí  5-7, ve svém kole máš tři možnosti: 1. útok 2. odpočinek 3.doplnění životů pokud máš džus")
    while zije and sliz_zije:
        vstup=oprava_int_input("1.útok 2.odpočinek 3.doplnění životů")
        if vstup==1 and energie > 4:
            energie-=5
            utok=random.randint(1,20)
            if "nůž" in inventar:
                utok+=5
            sliz_zivoty-=utok
            if sliz_zivoty>0:
                print("zaútočil jsi na protivníka a ubral jsi mu", utok, "životů, on teď má", sliz_zivoty, "životů, ty máš", zivoty, "životů a", energie, "energie")
            else:
                print("protivník umírá a ty vítězíš")
                souboj_chodba=0
                return
        elif vstup==1:
            print("nemáš dost energie, máš jen", energie, "energii, na útok potřebuješ 5")
            continue
        elif vstup==2:
            energie+=random.randint(5,7)
            print("odpočinul sis, teď máš", energie, "energie")
        elif vstup==3:
            if "lahev džusu (+ 50 životů)" in inventar:
                zivoty+=50
                if zivoty>100:
                    zivoty=100
                inventar.remove("lahev džusu (+ 50 životů)")
                print("přidal sis 50 životů, teď jich máš", zivoty)
            else:
                print("nemáš nic čím by sis mohl přidat životy")
                continue
        else:
            print("zadej 1, 2, nebo 3")
            continue
        if kolo==1:
            utok=random.randint(1, 15)
            zivoty-=utok
            if zivoty<1:
                zije=False
                print("protivník ti ubírá", utok, "životů, umíráš při souboji s hroudou slizu")
                zije=False
                return
            elif zivoty>0:
                print("protivník ti ubírá", utok, "životů, zbývá ti ", zivoty, ("životů"))
            kolo=0
        elif kolo==0:
            print("protivník odpočívá, máš", zivoty, "životů")
            kolo=1

#levá chodba, pokud je to první chodba do které hráč vstoupil tak proběhne souboj se slizem
def leva_chodba_def():
    global leva_chodba
    global start_mistnost
    global leva_mistnost
    global levy_suplik
    if souboj_chodba==1:
        souboj_sliz()
        if not zije:
            return
    print("rozhlížíš se po chodbě, vidíš skříň, fialové dveře a modré dveře")
    while zije:
        vstup=oprava_int_input("1. koukneš se do skříně 2. skrz fialové dveře 3. jdeš skrz modré dveře")
        if vstup==1:
            if levy_suplik ==1:
                inventar.append("kanistr")
                print("ve skříni nacházíš kanistr, máš u sebe", inventar)
                levy_suplik=0
            else:
                print("skříň je prázdná")
        elif vstup==2:
            print("jdeš skrz fialové dveře")
            leva_chodba=0
            leva_mistnost=1
            return
        elif vstup==3:
            print("jdeš skrz modré dveře")
            leva_chodba=0
            start_mistnost=1
            return
        else:
            print("napiš 1, 2 nebo 3")
#pravá chodba, pokud je to první chodba do které hráč vstoupil tak proběhne souboj se slizem
def prava_chodba_def():
    global start_mistnost
    global prava_chodba
    global pravy_suplik
    global prava_mistnost
    if souboj_chodba==1:
        souboj_sliz()
        return
    print("rozhlížíš se po chodbě, vidíš šuplík, oranžové dveře a červené dveře")
    while zije:
        vstup=oprava_int_input("1. koukneš se do šuplíku 2. skrz oranžové dveře 3. jdeš skrz červené dveře")
        if vstup==1:
            if pravy_suplik ==1:
                inventar.append("pojistka")
                print("v šuplíku nacházíš pojistku, máš u sebe", inventar)
                pravy_suplik=0
            else:
                print("šuplík je prázdný")
        elif vstup==2:
            print("jdeš skrz oranžové dveře do místnosti")
            prava_chodba=0
            prava_mistnost=1
            return
        elif vstup==3:
            print("jdeš skrz červené dveře")
            prava_chodba=0
            start_mistnost=1
            return
        else:
            print("napiš 1, 2 nebo 3")
#místnost ve které se dá do kanistru naplnit palivo
def leva_mistnost_def():
    global leva_chodba
    global leva_mistnost
    global prostredni_chodba
    print("rozhlížíš se po místnosti kde vidíš že ze stropu kape něco co smrdí a vypadá jako benzín taky vidíš fialové a zelené dveře")
    while True:
        vstup=oprava_int_input("1. jdeš ke kapajícímu palivu 2. jdeš skrz fialové dveře 3. jdeš skrz zelené dveře ")
        if vstup==1:
            if "kanistr" in inventar:
                print("palivo kape ze stropu a ty ho můžeš nabrat do kanistru")
                while True:
                    vstup=oprava_int_input("1. odcházíš od paliva 2. nabíráš palivo do kanistru")
                    if vstup==1:
                        break
                    if vstup==2:
                        inventar.remove("kanistr")
                        inventar.append("kanistr s palivem")
                        print("nabíráš palivo do kanistru, místo kanistru máš teď v inventáři kanistr s palivem, tvůj inventář:", inventar)
                        break
                    else:
                        print("zadej 1 nebo 2")
            elif"kanistr s palivem" in inventar:
                print("už sis tady naplnil kanistr, není tu nic k vidění")
            else:
                print("stojíš u kapajícího paliva, možná by se ti hodila nějaká nádoba na to aby sis ho nabral... odcházíš od kapajícího paliva")
                break
        elif vstup==2:
            leva_mistnost=0
            leva_chodba=1
            return
        elif vstup==3:
            print("vstupuješ skrz zelené dveře do chodby")
            leva_mistnost=0
            prostredni_chodba=1
            return
        else:
            print("zadej 1, 2 nebo 3 ")
#místnost ve které se dá opravit elektřina pokud má hráč pojistku
def prostredni_chodba_def():
    global elektrina
    global leva_mistnost
    global prostredni_chodba
    global suplik_mid_chodba
    global prava_mistnost
    print("rozhlížíš se a vidíš: display na stěně, šuplík, zelené dveře a žluté dveře ")
    while True:
        vstup=oprava_int_input("1. jdeš k displayi 2. jdeš k šuplíkům 3. jdeš skrz zelené dveře 4. skrz žluté dveře")
        if vstup==1:
            if elektrina==0:
                print("přicházíš k displayi, je na něm napsáno:,,primární zdroj: NEFUNKČNÍ, sekundární zdroj:POUŽIT,, vedle displaye je nějaká skříňka se znakem elektřiny")
                while True:
                    vstup=oprava_int_input("co uděláš 1. koukneš se do skříňky 2. odejdeš")
                    if vstup==1:
                        if "pojistka" in inventar:
                            print("ve skříňce je přepálená pojistka, vypadá jakože se dá nahradit pojistkou co máš u sebe")
                            while True:
                                vstup=oprava_int_input("1. nahradit pojistku 2. odejít")
                                if vstup==1:
                                    print("nahradil jsi pojistku, display říká: primární zdroj:funkční sekundární zdroj:nepoužívá se, odcházíš od skříňky")
                                    elektrina=1
                                    inventar.remove("pojistka")
                                    chybi.remove("opravit elektřinu")
                                    return
                                elif vstup==2:
                                    return
                                else:
                                    print("zadej 1 nebo 2")
                        else:
                            print("ve skříňce je vyhořelá pojistka, asi bys ji měl nahradit, ale nemáš jak, měl bys asi najít nějakou pojistku, odcházíš od displaye")
                            return
                    elif vstup==2:
                        return
                    else:
                        print("zadej 1 nebo 2")
            else:
                print("display hlásí: primární zdroj:funkční sekundární zdroj:nepoužívá se, není tu co dělat, odcházíš od skříňky ")
                return
        elif vstup==2:
            if suplik_mid_chodba==1:
                inventar.append("padák")
                print("v šuplíku jsi našel balík s nápisem ,,náhradní padák pro únikový modul,, teď máš v inventáři:", inventar)
                suplik_mid_chodba=0
                return
            else:
                print("v šuplíku nic není")
                return
        elif vstup==3:
            print("jdeš skrz zelené dveře do místnosti")
            prostredni_chodba=0
            leva_mistnost=1
            return
        elif vstup==4:
            print("jdeš skrz žluté dveře do místnosti")
            prostredni_chodba=0
            prava_mistnost=1
            return
        else:
            print("zadej 1, 2, 3 nebo 4")
#spojovací místnost, dá se skrz ní dostat do dvou chodeb a do únikového modulu
def prava_mistnost_def():
    global prava_mistnost
    global prava_chodba
    global prostredni_chodba
    global unikovy_modul
    print("rozhlížíš se a vidíš: žluté dveře, oranžové dveře, a dveře nad kterými je nápis ,,únikový modul,,")
    while True:
        vstup=oprava_int_input("kam jdeš: 1. žluté dveře 2. oranžové dveře 3. dveře s nápisem únikový modul")
        if vstup==1:
            print("procházíš skrz žluté dveře do chodby")
            prava_mistnost=0
            prostredni_chodba=1
            return
        elif vstup==2:
            print("procházíš skrz oranžové dveře do chodby")
            prava_mistnost=0
            prava_chodba=1
            return
        elif vstup==3:
            print("jdeš skrz dveře do únikového modulu")
            prava_mistnost=0
            unikovy_modul=1
            return
        else:
            print("zadej 1, 2 nebo 3")

#funkce která se ptá na odchod z modulu
def odchod_modul():
    global unikovy_modul
    global prava_mistnost
    while True:
        vstup=oprava_int_input("1. zůstaneš v místnosti 2. odcházíš z místnosti")
        if vstup==1:
            return
        elif vstup==2:
            print("odcházíš skrz dveře do místnosti")
            unikovy_modul=0
            prava_mistnost=1
            return
        else:
            print("zadej 1 nebo 2")
    

#funkce ve které se opravuje palivo, pokud ho hráč už opravil tak se vypíše že už opravit nemůže
def oprava_palivo():
    global palivo
    global unikovy_modul
    global prava_mistnost
    print("přicházíš k nálevu pro palivo")
    if "kanistr s palivem" in inventar and palivo==0 :
        while True:
            vstup=oprava_int_input("1.naliješ palivo do nálevky 2. odcházíš z místnosti")
            if vstup==1:
                inventar.remove("kanistr s palivem")
                chybi.remove("doplnit palivo")
                palivo=1
                print("nalil jsi palivo do nálevky, na displayi zmizel požadavek paliva")
                return
            elif vstup==2:
                print("odcházíš skrz dveře do místnosti")
                unikovy_modul=0
                prava_mistnost=1
                return
            else:
                print("zadej 1 nebo 2")
    if palivo==1:
        print("palivo už je doplněné, není co tu dělat")
        odchod_modul()
        return
    else:
        print ("nemáš co nalít do nálevky, někde na lodi se určitě dá sehnat palivo")
        odchod_modul()
        return
#funkce ve které se opravuje padák, pokud ho hráč už opravil tak se vypíše že už opravit nemůže
def oprava_padak():
    global padak
    global unikovy_modul
    global prava_mistnost
    print("přicházíš k úchytu pro padák")
    if "padák" in inventar :
        while True:
            vstup=oprava_int_input("1.dáš padák do úchytu 2. odcházíš z místnosti")
            if vstup==1:
                inventar.remove("padák")
                chybi.remove("opravit padák")
                padak=1
                print("dal jsi padák do úchytu, na displayi zmizel požadavek padáku")
                return
            elif vstup==2:
                print("odcházíš skrz dveře do místnosti")
                unikovy_modul=0
                prava_mistnost=1
                return
            else:
                print("zadej 1 nebo 2")
    if padak==1:
        print("padák už je opravený, na padáku není co opravit")
        odchod_modul()
        return
    else:
        print ("nemáš co do úchytu dát, ale někde tady na lodi určitě bude náhradní padák")
        odchod_modul()
        return  
#funkce ve které se vypisuje kde se opravuje elektřina, pokud ho hráč už opravil tak se vypíše že už opravit nemůže
def oprava_elektrina():
    if elektrina==0:
        print("display říká že elektřina se opravuje jinde než v této místnosti, prý v prostřední chodbě")
        odchod_modul()  
        return
    else:
        print("elektřina už je opravená, není tu co dělat")
        odchod_modul()  
        return
#funkce vypisuje co je třeba opravit a spouští funkce ve kterých se dají opravit
def unikovy_modul_def():
    global zije
    global unikovy_modul
    global prava_mistnost
    if elektrina==1 and palivo==1 and padak==1:
        print("display hlásí že vše je připraveno, svítí tlačítko start")
        while True:
            vstup=oprava_int_input("1. mačkáš tlačítko 2.odcházíš z místnosti")
            if vstup==1:
                print("modul se otřásá a odlétá na planetu Zemi, VYHRÁL JSI")
                zije=False
                return
            elif vstup==2:
                print("odcházíš skrz dveře do místnosti")
                unikovy_modul=0
                prava_mistnost=1
                return
            else:
                print("zadej 1 nebo 2")
    else :
        print("display v modulu hlásí že před odletem je třeba :", ", ".join(chybi))
        while True:
            vstup=oprava_int_input("1.jdeš k úchytu na padák 2. jdeš opravit elektřinu 3.jdeš doplnit palivo 4.odcházíš z místnosti")
            if vstup==1:
                oprava_padak()
                return
            elif vstup==2:
                oprava_elektrina()
                return
            elif vstup==3:
                oprava_palivo()
                return
            elif vstup==4:
                print("odcházíš skrz dveře do místnosti")
                unikovy_modul=0
                prava_mistnost=1
                return
            else:
                print("zadej 1, 2, 3 nebo 4")

print("v této hře budeš dostávat možnosti a budeš je vybírat pomocí možností 1,2,3... ")
print("můžeš kdykoliv napsat inventář nebo životy a dostaneš informace o svém inventáři a zdraví")
#tato část cykluje všechny místnosti a spouští místnost ve které jsi
while zije:
    #první část, spustí se jen na začátku.
    if probouzeni==1:
        probuzeni_def()
    if zije and start_mistnost==1:
        start_mistnost_def()
    if zije and leva_chodba==1:
        leva_chodba_def()
    if zije and prava_chodba==1:
        prava_chodba_def()
    if zije and leva_mistnost==1:
        leva_mistnost_def()
    if zije and prostredni_chodba==1:
        prostredni_chodba_def()
    if zije and prava_mistnost==1:
        prava_mistnost_def()
    if zije and unikovy_modul==1:
        unikovy_modul_def()
```
